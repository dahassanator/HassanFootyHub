```python
import pandas as pd
from statsbombpy import sb
from mplsoccer import VerticalPitch
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)  # so wide DataFrames don't get truncated
```

```python
competitions = sb.competitions()
print(competitions.shape)
competitions.head()
```

```python
ucl = competitions[competitions['competition_name'] == 'Champions League']
print(ucl.shape)
ucl    
```

```python
england = competitions[competitions['country_name'] == 'England']
print(england.shape)
england
```

```python
matches = sb.matches(competition_id=2, season_id=27)
print(matches.shape)
matches[['match_id', 'home_team', 'away_team', 'home_score', 'away_score', 'competition_stage']].head(10)
```

```python
events = sb.events(match_id=3754047)  # <- replace with a match_id from the table above
print(events.shape)
events['type'].value_counts()
```

```python
shots = events[events['type'] == 'Shot'].copy()
print(shots.shape)
```

```python
shot_cols = [
    'minute', 'player', 'team', 'shot_outcome',
    'shot_body_part', 'shot_type', 'shot_statsbomb_xg', 'location'
]
shots[shot_cols].head(10)
```

```python
print(shots['shot_outcome'].value_counts())
print()
print(shots['shot_body_part'].value_counts())
print()
print(shots['shot_type'].value_counts())
```

```python
shots[['x', 'y']] = pd.DataFrame(shots['location'].tolist(), index=shots.index)
shots[['x', 'y', 'shot_statsbomb_xg', 'shot_outcome']].head()
```

```python
pitch = VerticalPitch(pitch_type='statsbomb', half=True, line_color='black')
fig, ax = pitch.draw(figsize=(8, 8))

pitch.scatter(shots['x'], shots['y'], ax=ax, s=100, edgecolors='black')
plt.show()
```

```python
goals = shots[shots['shot_outcome'] == 'Goal']
misses = shots[shots['shot_outcome'] != 'Goal']

pitch = VerticalPitch(pitch_type='statsbomb', half=True, line_color='black')
fig, ax = pitch.draw(figsize=(8, 8))

# misses: size scaled by xG, hollow
pitch.scatter(misses['x'], misses['y'], ax=ax,
              s=misses['shot_statsbomb_xg'] * 500,
              edgecolors='grey', facecolors='none', label='No goal')

# goals: size scaled by xG, filled red
pitch.scatter(goals['x'], goals['y'], ax=ax,
              s=goals['shot_statsbomb_xg'] * 500,
              color='red', edgecolors='black', label='Goal')

ax.legend(loc='lower center')
plt.show()
```

```python
all_shots = []
for match_id in matches['match_id']:
    ev = sb.events(match_id=match_id)
    match_shots = ev[ev['type'] == 'Shot']
    all_shots.append(match_shots)
```

```python
shots_all = pd.concat(all_shots, ignore_index=True)
shots_all = shots_all.copy()
print(shots_all.shape)
```

```python
shots_all['is_goal'] = (shots_all['shot_outcome'] == 'Goal').astype(int)
print("Actual goal rate:", shots_all['is_goal'].mean().round(3))
print("Mean StatsBomb xG:", shots_all['shot_statsbomb_xg'].mean().round(3))
```
